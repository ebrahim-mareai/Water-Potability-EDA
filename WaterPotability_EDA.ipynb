defining all elements that data set incloude it:
pH: How acidic or basic the water is. 7 is neutral. 
Hardness: The minerals (like calcium and magnesium) dissolved in the water. 
Solids: All the tiny dissolved bits floating in the water.
Chloramines: A chemical added to kill germs and keep water safe as it flows through pipes.
Sulfate: A natural mineral from water and soil.
Conductivity: How well water carries electricity. 
Organic Carbon: Tiny natural bits of carbon in the water, like specks of decaying leaves or soil. 
Trihalomethanes (THMs): A byproduct that forms when chlorine meets natural organic matter. .
Turbidity: How cloudy or clear the water looks. 
Potability: The final answer.

```python
# firstly, pH vs Potability
import matplotlib.pyplot as plt # plotting library 
import seaborn as sns # library for better-looking graphs 

sns.set_theme(style="whitegrid") # background of graph 

plt.figure(figsize=(8, 6)) # new graph window 

# group by Potability and calculate average pH 
avg_ph = df.assign(ph_filled=df['ph'].fillna(df['ph'].median())).groupby('Potability')['ph_filled'].mean().reset_index()

# draw a bar chart 
sns.barplot(x='Potability', y='ph_filled', hue='Potability', data=avg_ph, palette='Set2', legend=False)

plt.title('Average pH Level vs Water Potability (Bar Chart)', fontsize=14) # add title
plt.xlabel('Water Safety Status', fontsize=12) # add a label to the bottom 
plt.ylabel('Average pH Level', fontsize=12) # add a label to the left Y-axis 
plt.xticks([0, 1], ['Not Potable (0)', 'Potable (1)']) # rename x-axis labels 
plt.show() # show graph
```

## the chart show:
the average solids concentration across these three pH categories. Based on the height of the bars:
Acidic water has the highest average solids concentration (tallest bar).
Neutral water has the second-highest concentration.
Basic water has the lowest average solids concentration among the three groups. 


```python
# third, Solids vs Potability
sns.set_theme(style="whitegrid") # background of graph

plt.figure(figsize=(8, 6)) # new graph window 

# Fill any missing values 
df['ph_filled'] = df['ph'].fillna(df['ph'].median())

# create pH categories from continuous values 
df['pH_Category'] = pd.cut(df['ph_filled'], bins=[0, 6, 8, 14], labels=['Acidic (<6)', 'Neutral (6-8)', 'Basic (>8)'], include_lowest=True)

# group by pH category and calculate average Solids 
avg_solids_by_ph = df.groupby('pH_Category', observed=True)['Solids'].mean().reset_index()

# draw a bar chart 
sns.barplot(x='pH_Category', y='Solids', hue='pH_Category', data=avg_solids_by_ph, palette='Set2', legend=False)

plt.title('Average Solids Level by pH Category (Bar Chart)', fontsize=14) # add title
plt.xlabel('pH Category (Acidity Level)', fontsize=12) # add a label to the bottom 
plt.ylabel('Average Solids Concentration', fontsize=12) # add a label to the left Y-axis 
plt.show() # show graph
```

# the nine scatter plots show:
the comparison of different water quality metrics against Potability. Based on the distribution of the data points:
The data points are densely clustered in two distinct horizontal lines (at 0 and 1), meaning the observations sit at the two potability values rather than forming a continuous trend.
Most metrics (such as pH, Hardness, and Solids) individually do not show a clear linear separation or obvious threshold between potable and non-potable water, since their values are widely distributed across both categories.
Overall, no single metric provides a clean visual boundary between potable and non-potable water, suggesting that potability is not determined by any one of these metrics alone.

```python
# white grid style for clean graphs
sns.set_theme(style="whitegrid")

# figure with 3x3 subplots (9 water quality elements)
fig, axes = plt.subplots(3, 3, figsize=(18, 15))

# flatten 3x3 axes to 1d for easy iteration
axes = axes.flatten()

# list of all water quality element columns
elements = ['ph', 'Hardness', 'Solids', 'Chloramines', 'Sulfate', 'Conductivity', 'Organic_carbon', 'Trihalomethanes', 'Turbidity']

# loop through each element + subplot axis
for idx, element in enumerate(elements):
    # scatter plot colored by potability (0 or 1), coolwarm palette
    scatter = axes[idx].scatter(df[element], df['Potability'], 
                                 c=df['Potability'], 
                                 cmap='coolwarm', 
                                 alpha=0.6, 
                                 s=20, 
                                 edgecolors='k', 
                                 linewidth=0.5)
    
    # subplot title
    axes[idx].set_title(f'{element} vs Potability', fontsize=12, fontweight='bold')
    
    # x-axis label
    axes[idx].set_xlabel(element, fontsize=10)
    
    # y-axis label
    axes[idx].set_ylabel('Potability', fontsize=10)
    
    # dashed line at y=0.5 (separates potable vs not)
    axes[idx].axhline(y=0.5, color='gray', linestyle='--', linewidth=1, alpha=0.7)

# hide unused subplots (only if fewer than 9 elements)
for idx in range(len(elements), len(axes)):
    axes[idx].set_visible(False)

# adjust spacing between subplots
plt.tight_layout(pad=3.0)

# main figure title (large, bold)
plt.suptitle('Comprehensive Water Quality Analysis: All Elements vs Potability\n(Entire Dataset: 3,276 Samples)', 
             fontsize=16, fontweight='bold', y=1.02)

# render figure
plt.show()
```

# the code show density plots comparing the distribution of various water quality metrics (like pH, Hardness, and Chloramines) :
for Potable (orange) and Not Potable (blue) water. The graphs reveal that the distribution curves for both categories highly overlap, indicating that these individual metrics do not provide a clear visual separation or distinct threshold between safe and unsafe water.

```python


# white grid style for clean graphs
sns.set_theme(style="whitegrid")

# figure with 5x2 grid (holds 9 element plots)
fig, axes = plt.subplots(5, 2, figsize=(16, 20))

# flatten 2d axes to 1d for easy looping
axes = axes.flatten()

# list of all 9 water quality element columns
elements = ['ph', 'Hardness', 'Solids', 'Chloramines', 'Sulfate', 'Conductivity', 'Organic_carbon', 'Trihalomethanes', 'Turbidity']

# loop through each element + subplot axis
for i, col in enumerate(elements):
    
    # density curve for "not potable" (potability == 0), blue filled
    sns.kdeplot(data=df[df['Potability'] == 0], x=col, ax=axes[i], color='blue', fill=True, alpha=0.4, label='Not Potable')
    
    # density curve for "potable" (potability == 1), orange filled
    sns.kdeplot(data=df[df['Potability'] == 1], x=col, ax=axes[i], color='orange', fill=True, alpha=0.4, label='Potable')
    
    # subplot title
    axes[i].set_title(f'Density Plot of {col}', fontsize=12, fontweight='bold')
    
    # x-axis label
    axes[i].set_xlabel(col, fontsize=10)
    
    # y-axis label
    axes[i].set_ylabel('Density', fontsize=10)
    
    # legend top-right
    axes[i].legend(loc='upper right')

# hide last empty subplot (9 elements, 10 spaces)
axes[-1].set_visible(False)

# adjust spacing between subplots
plt.tight_layout()

# main figure title (large, bold)
plt.suptitle('Density Distributions of Water Quality Elements by Potability (Entire Dataset)', 
             fontsize=16, fontweight='bold', y=1.01)

# render figure
plt.show()
```

# in this cell we will give final result for all nonpotable and potabl with fix missing values.
The bar chart illustrates the distribution of water samples according to their potability classification.
The number and percentage of potable and non-potable samples provide an overview of the target variable distribution. This comparison helps identify whether the dataset contains an imbalance between the two classes, which is relevant when interpreting subsequent analyses of water quality characteristics. However, the classification represents the labels provided in the dataset and does not independently establish whether a sample meets all applicable drinking-water safety standards.

```python

# Final Visualization: Potable vs Non-Potable Water

import matplotlib.pyplot as plt

# Count the samples in each potability category
potability_counts = df["Potability"].value_counts().reindex(
    [0, 1], fill_value=0
)

# Define labels and values
labels = ["Non-Potable Water", "Potable Water"]
counts = potability_counts.values
total = counts.sum()
percentages = (counts / total * 100) if total > 0 else [0, 0]

# Create the bar chart
fig, ax = plt.subplots(figsize=(9, 6))

bars = ax.bar(
    labels,
    counts,
    color=["#E76F51", "#2A9D8F"],
    edgecolor="black",
    width=0.55
)

# Display count and percentage above each bar
for bar, count, percentage in zip(bars, counts, percentages):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + total * 0.01,
        f"{count:,}\n({percentage:.1f}%)",
        ha="center",
        va="bottom",
        fontsize=12,
        fontweight="bold"
    )

# Add titles and labels
ax.set_title(
    "Distribution of Potable and Non-Potable Water Samples",
    fontsize=14,
    fontweight="bold",
    pad=15
)
ax.set_xlabel("Water Potability Category", fontsize=12)
ax.set_ylabel("Number of Samples", fontsize=12)

ax.set_ylim(0, max(counts) * 1.15 if max(counts) > 0 else 1)
ax.grid(axis="y", linestyle="--", alpha=0.3)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```
